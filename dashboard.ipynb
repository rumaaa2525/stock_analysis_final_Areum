import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

df = pd.read_csv('data/processed/data_scope_summary.csv')
df['date'] = pd.to_datetime(df['date'])

start_date = df['date'].min().strftime('%Y-%m-%d')
end_date = df['date'].max().strftime('%Y-%m-%d')
total_days = df['date'].nunique()
avg_stocks = int(df['valid_files'].mean())
total_missing = int(df['missing_files'].sum())
#| content: valuebox
#| title: "누적 결측 및 오류 파일 수"
dict(
    icon = "exclamation-triangle",
    color = "warning",
    value = f"{total_missing:,} 건"
)
fig1 = px.line(df, x='date', y='valid_files', 
              labels={'date': '수집 일자', 'valid_files': '정상 종목 수'},
              title='일자별 정상 수집 주식 종목 수 시계열')
fig1.update_traces(line_color='#007bff')
fig1.update_layout(margin=dict(l=20, r=20, t=40, b=20))
fig1.show()
total_valid_sum = df['valid_files'].sum()
total_missing_sum = df['missing_files'].sum()

fig2 = go.Figure(data=[go.Pie(
    labels=['정상 수집 HTML', '결측/오류 HTML'],
    values=[total_valid_sum, total_missing_sum],
    hole=.4,
    marker_colors=['#28a745', '#dc3545']
)])
fig2.update_layout(title_text='전체 아카이브 데이터 건전성', margin=dict(l=20, r=20, t=40, b=20))
fig2.show()
df_display = df.sort_values('date', ascending=False).head(10)
df_display['date'] = df_display['date'].dt.strftime('%Y-%m-%d')
df_display.columns = ['날짜', '전체 파일', '정상 파일', '결측/이상 파일']
df_display
